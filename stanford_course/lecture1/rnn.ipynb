# Recurrent Neural Networks 



RNNs can transform a sequence of data into vectors, vectors into seqeunces or sequences into sequences. 

The state of each time step of the sequence is dependent on the previous state of the sequence. 


```python
# represent a sequence as a vector first 
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
sentences = ["This is an example sentence", "Each sentence is converted to a vector"]

# Generate vectors (embeddings)
embeddings = model.encode(sentences)

print(embeddings[0]) 
```
